```python
from sympy import *


```

```python
x, a, t = symbols("x a t")
```

```python
f = x**2 * (x + cos(x)) - 4*x
f
```

```python
diff(f, x)
```

```python
y = Function("y")(x)
```

```python
ode = Eq((1 - x**2) * diff(y, x), 2*x*y)
display(ode)

```

```python
dsolve(ode, y)
```

```python
dsolve(ode, y, ics = {y.subs(x, 0): 1})
```
